```python
import pandas as pd

url = 'https://eds-217-essential-python.github.io/data/toolik_weather.csv'
toolik = pd.read_csv(url)

toolik[['Year', 'Month', 'Date', 'Daily_AirTemp_Mean_C']].head()
```

```python
# so what even is the type of a date column?
toolik['Date'].dtype

# so basically its an integer and its some giant unhelpful number
# but we can use certain functions to make it work nice
toolik['date'] = pd.to_datetime(toolik['Date'], format='%Y%m%d')

toolik[['Date', 'date']].head()
# now the type of date column is datetime64

```

```python
# lets go ahead and look at the first and last date in the data
print(toolik['date'].dtype)
print(toolik['date'].min())
print(toolik['date'].max())
```

```python
# you can pull a the month, day or year out of a date string using the 
# following commands
toolik['year'] = toolik['date'].dt.year
toolik['month'] = toolik['date'].dt.month
toolik['day'] = toolik['date'].dt.day

toolik[['date', 'year', 'month', 'day']].head()

#adding a day of year column
toolik['dayyear'] = toolik['date'].dt.dayofyear
toolik['dayyear']
```

```python
# now it is easy to find sum stats by month or day or etc
toolik.groupby('month')['Daily_AirTemp_Mean_C'].mean().round(2)

# and its a good idea to run the mean with a count stat at the same time
toolik.groupby('year')['Daily_AirTemp_Mean_C'].agg(['count', 'mean']).head(3).round(2)
```

```python
# now lets do an analysis if the arctic is actually getting colder
# as the years pass
early = toolik[toolik['year'] <= 1998]
late = toolik[toolik['year'] >= 2009]

# first we create two dataframes, one of the first 11 years of data and
# one of the last 11 years of data

comparison = pd.DataFrame({
    'early': early.groupby('month')['Daily_AirTemp_Mean_C'].mean(),
    'late': late.groupby('month')['Daily_AirTemp_Mean_C'].mean(),
})

comparison['change'] = comparison['late'] - comparison['early']

comparison.round(2)
#then we add both grouped data frames to a bigger df, and create a column that compares 
# the monthly mean of each group early and late

print(comparison['change'].mean())
```
