```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = 'https://eds-217-essential-python.github.io/data/penguins.csv'
penguins = pd.read_csv(url)
```

```python
penguins.shape
penguins.columns
penguins.dtypes
```

```python
penguins.groupby('species').count()
# penguins.groupby('island').count()
```

```python
penguins.isnull().sum()
penguins[penguins['sex'].isnull()]
```

```python
birds = penguins.dropna()
birds.groupby('species').count()
```

```python
# and here is a histogram in sns
plt.figure(figsize=(7, 5))

sns.histplot(data=birds, x='flipper_length_mm')

plt.xlabel('flipper length')
plt.title('penguin flipper')
plt.show()
```

```python
# and here is a histogram in sns
plt.figure(figsize=(7, 5))

sns.histplot(data=birds, x='flipper_length_mm', hue='species')

plt.xlabel('flipper length')
plt.title('penguin flipper')
plt.show()
```

```python
# and here is a histogram in sns
plt.figure(figsize=(7, 5))

sns.histplot(data=birds, x='body_mass_g', hue= 'species')

plt.xlabel('flipper length')
plt.title('penguin flipper')
plt.show()
```

```python
# and here is a histogram in sns
plt.figure(figsize=(7, 5))

sns.histplot(data=birds, x='body_mass_g')

plt.xlabel('flipper length')
plt.title('penguin flipper')
plt.show()
```

```python
# sns uses the column names as axis names, which isnt always great
# you can override this by specifying the axis names

sns.scatterplot(data=birds, x='flipper_length_mm', y='body_mass_g')
plt.xlabel('flipper size')
plt.ylabel('body mass')
```

```python
# sns uses the column names as axis names, which isnt always great
# you can override this by specifying the axis names

sns.scatterplot(data=birds, x='flipper_length_mm', y='body_mass_g', hue= 'species')
plt.xlabel('flipper size')
plt.ylabel('body mass')
```

```python
# sns uses the column names as axis names, which isnt always great
# you can override this by specifying the axis names

sns.scatterplot(data=birds, x='bill_length_mm', y='bill_depth_mm')
plt.xlabel('bill length')
plt.ylabel('bill depth')
```

```python
# sns uses the column names as axis names, which isnt always great
# you can override this by specifying the axis names

sns.scatterplot(data=birds, x='bill_length_mm', y='bill_depth_mm', hue= 'species')
plt.xlabel('bill length')
plt.ylabel('bill depth')
```

```python
birdavg = birds.groupby('species')['body_mass_g'].mean().sort_values(ascending= False)

plt.figure(figsize=(7, 5))

sns.barplot(x= birdavg.values, y= birdavg.index)

plt.xlabel('avg body mass')
plt.ylabel('species')
plt.title('which species is biggest?')
plt.show()
```

```python
birds.groupby('species')['body_mass_g'].agg(['count','mean'])

birdisland = birds['island'].value_counts()


plt.figure(figsize=(7, 5))

sns.barplot(x= birdisland.values, y= birdisland.index)

plt.xlabel('count')
plt.ylabel('island')
plt.title('birds per island')
plt.show()

birdisland
```
