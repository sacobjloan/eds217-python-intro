```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://eds-217-essential-python.github.io/data/toolik_weather.csv"
df = pd.read_csv(url)

df.describe()

```

```python
monthly = df.groupby('Month')
by_station = df.groupby('Station')

monthly_means = monthly[6].mean()
monthly_means

july_temp = 11.2
july_precip = 2.62
print(f"AtTollick in july it was {july_temp} and it rained {july_precip}")
```

```python
plt.plot(monthly_means)
plt.title("Toolik Monthly Temperatures")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
plt.bar(monthly_means.index, monthly_means)
plt.title("Average Temperature by Month")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.show()

warm = "12 degrees celcius"
print(f"Toolik's warmest month averages about {warm}")
```

```python
monthly_means.to_csv("monthly_means.csv", header=True, float_format="%.2f")
data = "monthly_means.csv"

print(f"i saved my data in {data}")
```
