```python
import pandas as pd

url = 'https://eds-217-essential-python.github.io/data/openaq_goleta_measurments.csv'
goleta = pd.read_csv(url)

pm25 = goleta[goleta['parameter'] == 'pm25'].copy()
pm25.shape
```

```python
pm25.sort_values('value').head() # default it sorts smallest to largest

pm25.sort_values('value', ascending=False).head() # this is how you make it go from largest to smallest

pm25.sort_values('parameter', ascending=False).head() # can even sort by strings

pm25.sort_values('datetimeUtc', ascending=False).head(10) # can even sort by strings

```

```python
# the top-N pattern is the following code
# df.sort_values('column', ascending=False).head(n)

# you can even choose which columns you want if you index at the end of the code

pm25.sort_values('value', ascending=False).head(10)[['datetimeLocal', 'value', 'unit']]

# double brackets is necessary [[]]

# you can filter, then sort in that order

goleta[goleta['parameter'] == 'pm10'].sort_values('value', ascending=False).head(5)[
    ['datetimeLocal', 'value', 'unit']
]
```

```python
# you can get the row # of a max using the following code

worsthour = pm25['value'].idxmax()

# that returns the row index, then you could use that to find a specfic cell

pm25.loc[worsthour,'datetimeLocal']

pm25['value'].max()
pm25[pm25['value'] == pm25['value'].max()]
```
