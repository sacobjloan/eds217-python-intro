## 1. A yes-or-no question


```python
temperature = 8.4
temperature > 0 
type(temperature > 0)  # has boolean type, which is 1 or 0 or true and false basically

# how to find specific row in a df
toolik.loc[0] # this indexes the first row of the df
toolik.loc[0:3] # this indexes the first 4 rows
toolik.loc[[1,3]] # this indexes the 2nd and 4th row
rowswanted = [1,3,4,6]
toolik.loc[rowswanted] # can index by list

toolik.loc[1:4, 'Year'] # can index by row and column

```

## 2. The comparison operators


```python
print(temperature > 0)
print(temperature < 0)
print(temperature >= 0)
print(temperature == 8.4)
# print(temperature = 8.4) this is busted bc the single equals is assignment, double is a comparison statement
print(temperature != 8.4) # this means not equal to

toolik.loc[0,'Station'] == 'TLKMAIN'




```

## 3. Asking two questions at once


```python
temperature
precipitation = 0.0

print(temperature > 0 and precipitation == 0) # use and to combine two statements
print(temperature > 0 or precipitation == 1) # use or to combine two statements in another way
print(not temperature >  0) # this creates the opposite of the conditional statement, negation opperator

# what about &? |? they are in other libraries so we will find them again
```

## 4. if / elif / else


```python
if temperature > 0 :
    print("above freezing")

if temperature > 0:
    print("above freezing")
else:
    print("below freezing")

temperature = -12

if temperature > 10:
    category = 'mild'
elif temperature > 0:
    category = 'cool'
elif temperature > -5:
    category = 'coldish'
else:
    category = 'cold'

print(category)
```

## 5. Asking a whole column


```python
# calling a section as a loop
for value in toolik.loc[[0,1,2], 'Daily_AirTemp_Mean_C']:
    if value > 7:
        print(f"{value} is warm")
    elif value > 0:
        print(f"{value} is mild")
    else:
        print(f"{value} is brrr!")

# doing a whole column at once

toolik['Daily_AirTemp_Mean_C'] > 0

toolik['Daily_Precip_Total_mm'] == 0

# now how do we combine the two statements above?

# use parenthesis to group the statements

# when using ands and ors we switch to &s and |s

(toolik['Daily_AirTemp_Mean_C'] > 0) & (toolik['Daily_Precip_Total_mm'] == 0) # now each true means the row satisfies BOTH conditions

```
