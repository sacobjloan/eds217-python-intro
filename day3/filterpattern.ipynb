```python
import pandas as pd

url = 'https://eds-217-essential-python.github.io/data/openaq_goleta_measurments.csv'
goleta = pd.read_csv(url)
goleta.shape
```

```python
goleta[goleta['value'] > 20] # this is filtering everything that has a number > 20 in the value column

badvalues = goleta[goleta['value'] < 0]   # these are all the rows that have a value <0, which makes no sense hwo do you have negative concentration
badvalues

ozone = goleta[goleta['parameter'] == 'o3'] # this gives us all the rows that are measuring o3 or ozone
ozone.head()
```

```python
goleta[goleta['parameter'] == 'pm25']['value'].describe()
# the above filters the df for columns where paremeter is pm25, then describes the 'value' column 

#pm25 analysis below, also filtering out values under 9
pm25anal = goleta[(goleta['parameter'] == 'pm25') & (goleta['value'] > 9)]
pm25anal.describe()
 
```

```python
# using a ~ to turn a condition around (same as !=)

not_ozone = goleta[~(goleta['parameter'] == 'o3')]

# the above returns all rows where paremeter is NOT o3

# same as this

goleta[goleta['parameter'] != 'o3'].shape
```

```python
# using .isin to filter a littler easier

particulates = goleta[goleta['parameter'].isin(['pm25', 'pm10'])]
particulates

# the above filters goleta by all rows that have pm25 or pm10 in the parameter column

# can do the above with a list which is nice and easy

particulateswecareabout = ['pm25','pm10','o3']
goleta[goleta['parameter'].isin(particulateswecareabout)]

particulates.columns

particulates[['unit', 'value']] # need double backets for some reason?
```

```python
#using the .copy thing to store data

o3df = goleta[goleta['parameter'] == 'o3'].copy()

goleta['value'].max()
goleta[goleta['value'] == goleta['value'].max()] # this is how you find a row with a specific value, like the max value in df

```
