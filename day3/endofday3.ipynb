```python
import pandas as pd
url = 'https://github.com/TheEconomist/banana-index-data/releases/download/1.0/bananaindex.csv'
data = pd.read_csv(url, index_col = 'entity')
```

```python
bad = ['year', 'Banana values', 'type', 'Chart?', 'Unnamed: 16']
data = data.drop(bad, axis = 'columns')
```

```python
data.shape
data.head()
data.info()


data.loc['Bananas', 'Bananas index (kg)']
```

```python
data.sort_values('Bananas index (kg)', ascending = False)
data.sort_values('Bananas index (1000 kcalories)', ascending = False)
data.sort_values('Bananas index (100g protein)', ascending = False)
```

```python
data['Bananas index (kg)'] > 10
data[data['Bananas index (kg)'] > 10].shape
data[~(data['Bananas index (kg)'] > 1)].shape
data[~(data['Bananas index (kg)'] > 1)]

```

```python
doublecond = data[(data['Bananas index (kg)'] > 5
) &
(data['Bananas index (100g protein)'] < 1
)].copy()

doublecond.sort_values('Bananas index (kg)', ascending = False)
```
