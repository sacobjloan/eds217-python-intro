```python
import pandas as pd

url = 'https://eds-217-essential-python.github.io/data/marine_microplastics.csv'
plastics = pd.read_csv(url)
plastics.shape
```

```python
plastics.columns.tolist() # classic code for checking column names
```

```python
plastics.isnull().sum() # this tells us how many empty cells are in a column
```

```python
# so there is some missing data, now we can look at them more closely
plastics.isnull().sum().sort_values(ascending=False).head(5)

# to drop rows with no data you can use .dropna()
plastics.dropna().shape

# BUT if we just do it with no modifiers we lose a lot of data
plastics.dropna().shape[0] / plastics.shape[0]
# we only have 3.6 percent of the data 
```

```python
# we can use subset=[] to be more precise with our .dropna()

measured = plastics.dropna(subset=['Measurement'])
measured.shape

plastics['Unit'].value_counts()
```

```python
# was that a good idea? idk, lets check the value counts for unit column and see what it looks like

plastics['Unit'].value_counts()

measured['Unit'].value_counts()

# all the pices/10 mins have vanished when dropping NAs from measurement column
```

```python
located = plastics.dropna(subset= 'Oceans')
# created a table without NAs in the Oceans column


```

```python
# now we learn how to fill empty cells with data
# of our choice

plastics['Regions'] = plastics['Regions'].fillna('Unspecified')
plastics['Regions'].value_counts().head(5)

plastics.isnull().sum().sort_values(ascending=False).head(3)
# now we see that Regions column is no longer in the top 3
# NA columns
```

```python
# what about duplicated rows?
plastics.duplicated().sum()
# above code sums up all the duplicated rows in a frame, but 
# only counts rows if all columns are identical

#to check if a there are duplicates in certain rows, you
# can subset ofc

sample_columns = ['Latitude', 'Longitude', 'Date', 'Measurement', 'Unit']
# create a list of columns that have actual data
plastics.duplicated(subset=sample_columns).sum() # subset by list
```

```python
# to drop duplicates you can use drop_duplicates(subset=)

deduplicated = plastics.drop_duplicates(subset=sample_columns)
deduplicated.shape

(15389, 22)
```

```python
# its good to check the repeats and make sure they really are
# repeats
repeats = plastics[plastics.duplicated(subset=sample_columns, keep=False)]
print(repeats.shape)
repeats[['Measurement', 'Unit', 'Sampling Method', 'Short Reference']].head(6)
```

```python
# maybe you want to change the type of a column in a df
plastics['Accession Number'] = plastics['Accession Number'].astype(str)
plastics['Accession Number'].head(3)

plastics['OBJECTID'] = plastics['OBJECTID'].astype(str)
plastics['OBJECTID'].dtype
```

```python
# heres what a concise and commented out cleaning looks like
clean = pd.read_csv(url)

# Open-ocean samples have no named region or subregion; that is information, not a gap.
clean['Regions'] = clean['Regions'].fillna('Unspecified')
clean['SubRegions'] = clean['SubRegions'].fillna('Unspecified')

# Accession numbers are labels, not quantities.
clean['Accession Number'] = clean['Accession Number'].astype(str)

# Keep only samples that have a measurement. This drops all 5,792 'pieces/10 mins'
# rows, which is acceptable here because we are about to average measurements.
clean = clean.dropna(subset=['Measurement'])

clean.shape
```
