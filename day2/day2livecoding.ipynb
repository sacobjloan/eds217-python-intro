## 1. Making lists


```python
names = ['jacob','erin', 'gita']
names[0:2]
```

## 2. Getting things out of a list

```python
for name in names:
    print(f"{name} is a great student")
```

## 3. Changing a list


```python
names[0] = 'liam'
newstudents = ['somebody','anotherperson']
names + newstudents

names[:2] = newstudents
names
```

## 4. Looping over a list


```python
for name in names:
    print(f"{name} is a great student")
```

## 5. Making dictionaries


```python
names_pets = {'Nina':1, 'Prescilla':7,'Elyse':5, 'karolina':0}
names_pets['karla'] = 1
names_pets
```

## 6. Getting things out of a dictionary


```python
names_pets['Nina']
```

## 7. Looping over a dictionary

```python
for key, value in names_pets.items():
    print(f"{key} has {value} pets")
    if value > 1:
        print(f"{key} has too many pets!")
```
